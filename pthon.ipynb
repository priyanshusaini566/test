```python
num=int(input("Enter the number : "))
count=0
rem=0

while(num>0):
    rem=num%10
    count=count+rem
    num=num//10

print("the sum of digits of the given number will be : ",count)
```

```python
num=int(input("Enter the number : "))
count=0
rem=0

while(num>0):
    rem=num%10
    count=count*10+rem
    num=num//10

print("the reverse of digits of the given number will be : ",count)
```

```python
num=int(input("Enter the number : "))
n=num
count=0
rem=0

while(num>0):
    rem=num%10
    count=count*10+rem
    num=num//10

if(n==count):
    print("the given number is a palindrome")

else:
    print("the given number is not a palindrome")
```
